```python
df_asia=df[df["origin"] == "Asia"]
df_asia.head()
```

```python
df_asia = df_asia[["vehicle_weight", "model_year"]]
df_asia.head(7)
```

```python
x=df_asia.iloc[:7].to_numpy()
x
```

```python
xtx=np.dot(x.T, x)
xtx
```

```python
xtx_inv=np.linalg.inv(xtx)
xtx_inv
```

```python
y=[1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w=np.dot(np.dot(xtx_inv, x.T),y)
w
```
